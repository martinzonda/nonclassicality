```python
def Pjaycum(n,nu,tau,pe):
    nuf1 = nu**n/(1+nu)**(1+n)
    nuf2 = (1+nu)/nu
    nuf3 = nu/(1+nu)
    resA = (np.cos(tau*np.sqrt(n+1)))**2 + nuf2*(np.sin(tau*np.sqrt(n)))**2
    resB = (np.cos(tau*np.sqrt(n)))**2 + nuf3*(np.sin(tau*np.sqrt(n+1)))**2
    res = nuf1*(pe*resA + (1.0-pe)*resB)
    return res
```

```python
def p1_0fun(x):
    return -np.log(x)*x

def p2_1fun(x):
    return -0.5*x*(sp.special.lambertw(-x))

def p2_0fun(x):
    xc=1.59362426004004
    Pxc=np.exp(-xc)
    Cxc=xc**(n-1)*Pxc/factorial(n-1)
    y=np.where(x<Pxc,x*(-np.log(x))**n,Cxc*(1-x))
    return y

def pN_0fun(x,n):
    xc=n+sp.special.lambertw(-np.exp(-n)*n)
    Pxc=np.exp(-xc)
    Cxc=xc**(n-1)*Pxc/factorial(n-1)
    y=np.where(x<Pxc,x*(-np.log(x))**n/factorial(n),Cxc*(1-x))
    return y
```

```python
import math

def boundary_km_curve(k, m, mu_max=20, num_points=2000, remove_turning=True):
    """
    Generate the boundary curve defined by Eq. (2pk) via Poisson parametrization.

    Returns:
        x = P_k(mu)
        y = P_m(mu)

    Parameters:
    -----------
    k, m : int
        Photon indices (must satisfy m > k)
    mu_max : float
        Maximum value of mu
    num_points : int
        Number of sampling points
    remove_turning : bool
        If True, sorts points by x to remove loops (useful for convex envelope plots)

    Returns:
    --------
    x : np.ndarray
        Values of P_k
    y : np.ndarray
        Values of P_m
    """

    if m <= k:
        raise ValueError("Require m > k")

    mu = np.linspace(1e-8, mu_max, num_points)

    # Poisson probabilities
    x = np.exp(-mu) * mu**k / math.factorial(k)
    y = np.exp(-mu) * mu**m / math.factorial(m)

    # Remove numerical noise
    mask = (x > 1e-15) & (y > 1e-15)
    x = x[mask]
    y = y[mask]

    # Optional: enforce monotonic ordering (removes loops visually)
    if remove_turning:
        idx = np.argsort(x)
        x = x[idx]
        y = y[idx]

    return x, y
```
